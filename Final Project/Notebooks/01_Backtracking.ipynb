```python
# تعريف المشكلة بناءً على وصف المشروع
bin_capacity = 8 # سعة الصندوق القصوى [cite: 37]

# أحجام العناصر اللي محتاجين نوزعها في الصناديق [cite: 38-58]
items = [1, 2, 3, 4, 1, 2, 4, 3, 5, 3, 4, 4, 5, 1, 5, 4, 4, 6, 2, 7, 4]

print(f"عدد العناصر اللي معانا: {len(items)} عنصر")
print(f"السعة القصوى لكل صندوق: {bin_capacity}")

# خطوة لتسهيل الحل: ترتيب العناصر تنازلياً (من الأكبر للأصغر)
# ده بيساعد الـ Backtracking يخلص العناصر الكبيرة الأول ويوفر وقت (Pruning مبدئي)
items.sort(reverse=True)
print(f"العناصر بعد الترتيب: {items}")
```

```python
import time

# المتغيرات اللي هنحفظ فيها أحسن حل نوصلوله
best_num_bins = float('inf')
best_solution = []

def backtrack(item_index, current_bins, bin_assignments):
    global best_num_bins, best_solution
    
    # 1. Pruning (القص): لو عدد الصناديق الحالية بيساوي أو عدى أحسن حل، نوقف المسار
    if len(current_bins) >= best_num_bins:
        return
    
    # 2. Base Case (نقطة التوقف): لو خلصنا كل العناصر
    if item_index == len(items):
        best_num_bins = len(current_bins)
        # بناخد نسخة من التوزيعة الحالية عشان نحفظها
        best_solution = [list(bin_items) for bin_items in bin_assignments]
        return
    
    current_item = items[item_index]
    
    # 3. نجرب نحط العنصر في الصناديق المفتوحة حالياً
    for i in range(len(current_bins)):
        # لو الصندوق فيه مساحة كافية
        if current_bins[i] >= current_item:
            # نحط العنصر (بنقلل المساحة الفاضية وبنضيفه في القائمة)
            current_bins[i] -= current_item
            bin_assignments[i].append(current_item)
            
            # نكمل للعنصر اللي بعده (Recursion)
            backtrack(item_index + 1, current_bins, bin_assignments)
            
            # التراجع (Backtrack): نشيل العنصر تاني عشان نجرب احتمالات تانية
            current_bins[i] += current_item
            bin_assignments[i].pop()
            
    # 4. نجرب نفتح صندوق جديد للعنصر ده
    if len(current_bins) + 1 < best_num_bins:
        current_bins.append(bin_capacity - current_item)
        bin_assignments.append([current_item])
        
        backtrack(item_index + 1, current_bins, bin_assignments)
        
        # التراجع: نقفل الصندوق اللي فتحناه
        current_bins.pop()
        bin_assignments.pop()

# تشغيل الخوارزمية وحساب الوقت
print("جاري البحث عن أفضل توزيعة (Optimal Solution)...")
start_time = time.time()

backtrack(0, [], [])

end_time = time.time()

print("-" * 40)
print(f"أحسن عدد صناديق وصلنا له: {best_num_bins}")
print(f"الوقت المستغرق: {end_time - start_time:.4f} ثانية")
print("تفاصيل التوزيعة:")
for i, b in enumerate(best_solution):
    print(f"صندوق {i+1}: {b} (مجموعهم {sum(b)} من أصل {bin_capacity})")
```
